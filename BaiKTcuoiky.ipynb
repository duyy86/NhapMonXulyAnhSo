```
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt


image_files = ['pagoda.jpg', 'bird.jpg', 'halong.jpg']


def show_image(title, image, is_gray=False):
    plt.figure(figsize=(5, 4))
    if is_gray:
        plt.imshow(image, cmap='gray')
    else:
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.axis('off')
    plt.show()


for file in image_files:
    image = cv2.imread(file)
    if image is None:
        print(f"⚠️ Không thể đọc ảnh: {file}")
        continue

    base_name = os.path.splitext(os.path.basename(file))[0]

    median_filtered = cv2.medianBlur(image, 5)
    fname1 = f'{base_name}_median.jpg'
    cv2.imwrite(fname1, median_filtered)
    show_image(f'{base_name} - Median Filter', median_filtered)

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    sobel = cv2.magnitude(sobelx, sobely)
    sobel = cv2.convertScaleAbs(sobel)
    fname2 = f'{base_name}_sobel.jpg'
    cv2.imwrite(fname2, sobel)
    show_image(f'{base_name} - Sobel Filter', sobel, is_gray=True)

    swapped = image[:, :, [1, 2, 0]]
    fname3 = f'{base_name}_swapped.jpg'
    cv2.imwrite(fname3, swapped)
    show_image(f'{base_name} - GRB Swapped', swapped)

    lab_image = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
    L, A, B = cv2.split(lab_image)
    cv2.imwrite(f'{base_name}_L.jpg', L)
    cv2.imwrite(f'{base_name}_A.jpg', A)
    cv2.imwrite(f'{base_name}_B.jpg', B)
    show_image(f'{base_name} - L channel (LAB)', L, is_gray=True)
    show_image(f'{base_name} - A channel (LAB)', A, is_gray=True)
    show_image(f'{base_name} - B channel (LAB)', B, is_gray=True)

    print(f"✅ Đã xử lý và hiển thị ảnh: {file}")
```

Các biến mà bạn xác định trong một ô có thể dùng trong các ô khác sau này:

```
import cv2
import numpy as np
import random
import os
import matplotlib.pyplot as plt


image_files = ['bird.jpg', 'halong.jpg', 'pagoda.jpg']


def random_odd_kernel():
    return random.choice([3, 5, 7, 9, 11])


def show_image(title, img_path, is_gray=False):
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE if is_gray else cv2.IMREAD_COLOR)
    if img is None:
        print(f"❌ Không thể hiển thị ảnh: {img_path}")
        return

    plt.figure(figsize=(6, 4))
    if is_gray:
        plt.imshow(img, cmap='gray')
    else:
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img_rgb)

    plt.title(title)
    plt.axis('off')
    plt.show()


def process_images(method_key):
    for file in image_files:
        image = cv2.imread(file)
        if image is None:
            print(f"⚠️ Không thể đọc ảnh: {file}")
            continue

        base_name = os.path.splitext(os.path.basename(file))[0]
        is_gray = False  # Mặc định ảnh có màu

        if method_key == 'B':  # Gaussian Blur
            ksize = random_odd_kernel()
            print(f"[Gaussian] kernel = {ksize}")
            result = cv2.GaussianBlur(image, (ksize, ksize), 0)
            label = f"Gaussian {ksize}"
            fname = f'result_blur_{base_name}.jpg'

        elif method_key == 'M':  # Median Blur
            ksize = random_odd_kernel()
            print(f"[Median] kernel = {ksize}")
            result = cv2.medianBlur(image, ksize)
            label = f"Median {ksize}"
            fname = f'result_median_{base_name}.jpg'

        elif method_key == 'F':  # Bilateral Filter
            d = random.randint(5, 15)
            sigmaColor = random.randint(50, 150)
            sigmaSpace = random.randint(50, 150)
            print(f"[Bilateral] d={d}, sigmaColor={sigmaColor}, sigmaSpace={sigmaSpace}")
            result = cv2.bilateralFilter(image, d, sigmaColor, sigmaSpace)
            label = f"Bilateral d={d}"
            fname = f'result_bilateral_{base_name}.jpg'

        elif method_key == 'E':  # Canny
            t1 = random.randint(50, 100)
            t2 = random.randint(100, 150)
            print(f"[Canny] threshold1={t1}, threshold2={t2}")
            gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            result = cv2.Canny(gray, t1, t2)
            is_gray = True
            label = f"Canny {t1}-{t2}"
            fname = f'result_canny_{base_name}.jpg'

        elif method_key == 'R':  # Erosion
            k = random.randint(2, 5)
            print(f"[Erosion] kernel size = {k}")
            kernel = np.ones((k, k), np.uint8)
            result = cv2.erode(image, kernel, iterations=1)
            label = f"Erosion {k}x{k}"
            fname = f'result_erosion_{base_name}.jpg'

        elif method_key == 'D':  # Dilation
            k = random.randint(2, 5)
            print(f"[Dilation] kernel size = {k}")
            kernel = np.ones((k, k), np.uint8)
            result = cv2.dilate(image, kernel, iterations=1)
            label = f"Dilation {k}x{k}"
            fname = f'result_dilation_{base_name}.jpg'

        else:
            print("⚠️ Phím không hợp lệ.")
            return


        if not is_gray:
            cv2.putText(result, label, (10, 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

        cv2.imwrite(fname, result)
        print(f"✅ Đã lưu: {fname}")
        show_image(f"{label} - {file}", fname, is_gray)

def show_menu():
    print("===== MENU XỬ LÝ ẢNH =====")
    print("B: Gaussian Blur")
    print("M: Median Blur")
    print("F: Bilateral Filter")
    print("E: Canny Edge Detection")
    print("R: Erosion")
    print("D: Dilation")
    print("==========================")

    key = input("Nhập phím tương ứng để xử lý: ").strip().upper()
    if key in ['B', 'M', 'F', 'E', 'R', 'D']:
        process_images(key)
    else:
        print("❌ Phím không hợp lệ!")


if __name__ == "__main__":
    show_menu()
```

```
import cv2
import numpy as np
import random
import matplotlib.pyplot as plt

# Hàm hiển thị ảnh (RGB)
def show_image(title, img_path):
    image = cv2.imread(img_path)
    if image is None:
        print(f"❌ Lỗi đọc ảnh: {img_path}")
        return
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.imshow(image_rgb)
    plt.title(title)
    plt.axis('off')
    plt.show()

# === Ảnh 1: Thêm viền 35px mỗi cạnh ===
def add_margin(image_path, output_path):
    image = cv2.imread(image_path)
    if image is None:
        print(f"⚠️ Lỗi đọc ảnh: {image_path}")
        return
    padded = cv2.copyMakeBorder(image, 35, 35, 35, 35, cv2.BORDER_REFLECT)
    cv2.imwrite(output_path, padded)
    show_image("1️⃣ Thêm viền 35px", output_path)

# === Ảnh 2: Xoay 135 độ + Lật ngang ===
def rotate_135_flip(image_path, output_path):
    image = cv2.imread(image_path)
    if image is None:
        print(f"⚠️ Lỗi đọc ảnh: {image_path}")
        return
    h, w = image.shape[:2]
    center = (w // 2, h // 2)
    matrix = cv2.getRotationMatrix2D(center, 135, 1.0)
    rotated = cv2.warpAffine(image, matrix, (w, h))
    flipped = cv2.flip(rotated, 1)
    cv2.imwrite(output_path, flipped)
    show_image("2️⃣ Xoay 135° + Lật ngang", output_path)

# === Ảnh 3: Phóng to 5x + Gaussian blur + Điều chỉnh độ sáng và tương phản ===
def upscale_blur_and_adjust(image_path, output_path):
    image = cv2.imread(image_path)
    if image is None:
        print(f"⚠️ Lỗi đọc ảnh: {image_path}")
        return

    # Tăng kích thước 5 lần
    h, w = image.shape[:2]
    upscaled = cv2.resize(image, (w * 5, h * 5), interpolation=cv2.INTER_LINEAR)

    # Làm mịn bằng Gaussian blur 9x9
    blurred = cv2.GaussianBlur(upscaled, (9, 9), 0)

    # Điều chỉnh độ sáng và độ tương phản
    alpha = random.uniform(0.5, 2.0)
    beta = random.randint(-50, 50)
    adjusted = cv2.convertScaleAbs(blurred, alpha=alpha, beta=beta)

    cv2.imwrite(output_path, adjusted)
    print(f"📈 Đã điều chỉnh ảnh với α = {alpha:.2f}, β = {beta}")
    show_image("3️⃣ Phóng to + Blur + Sáng/Tương phản", output_path)

# === Đường dẫn 3 ảnh đầu vào ===
image1 = "bird.jpg"
image2 = "halong.jpg"
image3 = "pagoda.jpg"

# === Gọi các hàm xử lý ===
add_margin(image1, "output_image1_margin.jpg")
rotate_135_flip(image2, "output_image2_rotated_flipped.jpg")
upscale_blur_and_adjust(image3, "output_image3_final.jpg")

print("🎯 Đã xử lý hoàn tất tất cả yêu cầu của Câu 3.")
```
